```python
# Cell 2: โค้ดสร้างและแสดงกราฟ
import os
from pyvis.network import Network
from IPython.display import display, HTML

# นำเข้าคลาส KnowledgeGraph ของคุณ
from knowledge_graph import KnowledgeGraph 

def show_graph_in_notebook(db_path="./movex_kg.db"):
    print("กำลังโหลดข้อมูลและสร้างกราฟ...")
    kg = KnowledgeGraph.load_from_db(db_path)
    nx_graph = kg.graph
    
    # สร้าง PyVis Network (ตั้งค่า notebook=True สำหรับแสดงใน .ipynb)
    net = Network(
        height="600px",         # ปรับความสูงของกรอบกราฟได้ตามต้องการ
        width="100%", 
        bgcolor="#ffffff", 
        font_color="black", 
        directed=True,
        notebook=True,          # สำคัญ: บอกว่ารันใน jupyter notebook
        cdn_resources='remote'  # สำคัญ: โหลดสคริปต์จากเน็ตเพื่อไม่ให้พังใน Jupyter
    )

    # ชุดสี
    color_map = {
        "Product": "#f56b42",             # สีส้มแดง
        "Series": "#42b6f5",              # สีฟ้า
        "Material": "#9cd94c",            # สีเขียว
        "ManufacturingProcess": "#c5a382" # สีน้ำตาลอ่อน
    }

    # 1. สร้าง Nodes
    for node_id, node_data in nx_graph.nodes(data=True):
        node_type = node_data.get("node_type", "Unknown")
        label = node_id.split(":")[-1] 
        
        # ใส่ Tooltip (ข้อมูลเวลาเอาเมาส์ชี้)
        tooltip_text = f"Type: {node_type}\n"
        for k, v in node_data.items():
            if k != "node_type" and v:
                tooltip_text += f"{k}: {v}\n"

        net.add_node(
            node_id, 
            label=label, 
            title=tooltip_text, 
            color=color_map.get(node_type, "#cccccc"),
            group=node_type,
            size=15 if node_type == "Product" else 25
        )

    # 2. สร้าง Edges
    for source, target, edge_data in nx_graph.edges(data=True):
        relation = edge_data.get("relation", "")
        net.add_edge(
            source, 
            target, 
            title=relation,
            label=relation,
            color="#aaaaaa",
            arrows="to"
        )

    # เปิด Physics เด้งดึ๋ง
    net.toggle_physics(True)

    # คำสั่งโชว์ใน Output Cell ทันที
    return net.show("movex_graph.html")

# เรียกใช้งานฟังก์ชัน
show_graph_in_notebook()
```
