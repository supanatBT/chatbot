```python
from typing import List, Optional
from pydantic import BaseModel, Field
from llama_cloud_services import LlamaExtract

# 1. นิยาม Schema สำหรับการดึงข้อมูลดิบจากตาราง (Raw Data)
# ออกแบบมาเพื่อจัดการกับ Merged Cells ในตาราง Sprocket โดยเฉพาะ
class SprocketRaw(BaseModel):
    Art_Nr: str = Field(description="Article Number (e.g., 58901, 58801).")
    Z_Teeth: int = Field(description="Number of teeth (Z). Repeat this value for all rows in the same group.")
    Bore_mm: float = Field(description="Bore diameter in mm.")
    PD_mm: float = Field(description="Pitch Diameter in mm. Repeat for all rows in the group.")
    OD_mm: float = Field(description="Outside Diameter in mm. Repeat for all rows in the group.")
    S_mm: float = Field(description="Hub width or Thickness (S). Repeat for all rows in the group.")
    Material: str = Field(default="Polyamide", description="Material of the sprocket.")
    Sprocket_Type: str = Field(description="Identify if it is 'Drive' or 'Idler' sprocket based on the table header.")

class SprocketList(BaseModel):
    sprockets: List[SprocketRaw] = Field(description="List of extracted sprockets from the table.")

# 2. นิยาม Schema สำหรับข้อมูลที่ต้องการให้ AI ช่วยวิเคราะห์ (Enriched Data)
class SprocketAIInsights(BaseModel):
    compatible_chains: str = Field(description="Chain series this sprocket works with (e.g., 83 MFM PRO).")
    usage_context: str = Field(description="Detailed usage (e.g., Used as a drive sprocket for high-speed bottling lines).")
    technical_summary: str = Field(description="A short technical pitch for this sprocket.")

# 3. โครงสร้างสุดท้ายสำหรับนำไปลงฐานข้อมูล (Final Database Format)
class SprocketFinal(SprocketRaw):
    product_type: str = "Sprocket"
    series: str = Field(description="Sprocket Series Name (e.g., MF PRO Dedicated Sprockets)")
    ai_insights: SprocketAIInsights
    search_content: str = Field(description="Optimized text for semantic search.")

# --- เริ่มการทำงาน ---

extractor = LlamaExtract()

# สร้าง Agent สำหรับการดึงข้อมูล
agent = extractor.create_agent(
    name="Sprocket-Table-Parser-V5",
    data_schema=SprocketList
)

# สั่ง Extraction (ระบุ Path ของรูป table-7-7)
image_path = "/content/drive/MyDrive/Senior Project-20260109T082330Z-1-001/Senior Project/TOB/data/extracted_data_4/882_split_drive.jpg"
result = agent.extract(image_path)
```
